```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
std = StandardScaler()
lab = LabelEncoder()
ohe = OneHotEncoder(drop='first')
col = ColumnTransformer([('one hot',ohe,[i for i in categ if i != 'model']),
                        ('scalar',std,[i for i in num if i != 'selling_price'])],
                        remainder= 'passthrough')
```

```python
x['model'] = lab.fit_transform(x['model'])
```

```python
x = col.fit_transform(x)
```

```python
x = pd.DataFrame(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=45,)
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestRegressor,AdaBoostRegressor,GradientBoostingRegressor
from sklearn.linear_model import LinearRegression,RidgeCV,LassoCV
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error,r2_score,mean_absolute_error

```

```python
def evalu(pred,truth):
    print("rmse: ",root_mean_squared_error(truth,pred))
    print("r2 score: ",r2_score(truth,pred))
    print("mae: ",mean_absolute_error(truth,pred))
    


```

```python
from xgboost import XGBRegressor
```

```python
models = {
    "RandomForest" : RandomForestRegressor(),
    "LinearRegression" : LinearRegression(),
    "Ridge regression" : RidgeCV(),
    "Lasso regression" : LassoCV(),
    "k neighbors" : KNeighborsRegressor(n_neighbors=2),
    "Decision regressor" : DecisionTreeRegressor(),
    "Addaboost" : AdaBoostRegressor(n_estimators = 60, loss = 'linear'),
    "Gradient" : GradientBoostingRegressor(n_estimators = 100, loss = 'squared_error', learning_rate = 0.3, criterion = 'squared_error'),
    "xgboost" : XGBRegressor()
}
```

```python
for i in range (0,len(list(models))):
    model = list(models.values())[i]
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    print("*" * 35)
    print("\n")
    print(model)
    evalu(y_pred,y_test)
```

```python
knn_params = {"n_neighbors": [2, 3, 10, 20, 40, 50]}
add_params = {"loss" : ['linear', 'square', 'exponential'] , "n_estimators" : [50,60,70,80]}
grad_param = {'loss' : ['squared_error', 'absolute_error', 'huber', 'quantile'], "learning_rate" : [0.1,0.01,0.05,0.3],
              "n_estimators" : [50,80,100], "criterion" : ['friedman_mse', 'squared_error']}
```

```python
randomcv = [('KNN',KNeighborsRegressor(),knn_params),
            ('addaboost',AdaBoostRegressor(),add_params),
            ('gradboost',GradientBoostingRegressor(),grad_param)]
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model_p = {}
for name,model,param in randomcv:
    gc = RandomizedSearchCV(estimator=model,param_distributions=param,n_jobs=-1,
                                verbose=2,cv=3,n_iter=100,random_state=89)
    gc.fit(x_train,y_train)
    model_p[name] = gc.best_params_

for i in model_p:
    print(f"Best params for {i} -------")
    print(model_p[i])
```

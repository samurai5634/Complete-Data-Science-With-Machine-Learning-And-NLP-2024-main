```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
std = StandardScaler()
lab = LabelEncoder()
ohe = OneHotEncoder(drop='first')
col = ColumnTransformer([('one hot',ohe,[i for i in categ if i != 'model']),
                        # ('label',lab,['model']),
                        ('scalar',std,num)],
                        remainder= 'passthrough')
```

```python
x['model'] = lab.fit_transform(x['model'])
```

```python
x = col.fit_transform(x)
```

```python
x = pd.DataFrame(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=45,)
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression,RidgeCV,LassoCV
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import root_mean_squared_error,r2_score,mean_absolute_error

```

```python
def evalu(pred,truth):
    print("rmse: ",root_mean_squared_error(truth,pred))
    print("r2 score: ",r2_score(truth,pred))
    print("mae: ",mean_absolute_error(truth,pred))
    


```

```python
models = {
    "RandomForest" : RandomForestRegressor(),
    "LinearRegression" : LinearRegression(),
    "Ridge regression" : RidgeCV(),
    "Lasso regression" : LassoCV(),
    "k neighbors" : KNeighborsRegressor(),
    "Decision regressor" : DecisionTreeRegressor()
}
```

```python
models.val
```

```python
for i in range (0,len(list(models))):
    model = list(models.values())[i]
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    print("*" * 35)
    print("\n")
    print(model)
    evalu(y_pred,y_test)
```

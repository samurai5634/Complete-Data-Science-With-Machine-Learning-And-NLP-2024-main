```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import pandas as pd
```

```python
x,y = make_blobs(n_samples=1000,n_features=2,centers=3)
```

```python
x
```

```python
plt.scatter(x[:,0],x[:,1],c = y)
```

```python
from sklearn.preprocessing import StandardScaler
std = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=56,)
```

```python
x_train_scaled = std.fit_transform(x_train)
x_test_scaled = std.transform(x_test)
```

```python
from sklearn.cluster import KMeans
```

```python
### Elbow method
wcss = []
for k in range(1,15):
    kmeans = KMeans(init= "k-means++",n_clusters= k)
    kmeans.fit_transform(x_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,15), wcss)
plt.xticks(range(1,15))
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
```

```python
kmeans = KMeans(init= "k-means++",n_clusters= 3)
```

```python
kmeans.fit_predict(x_train_scaled)
```

```python
y_pred  = kmeans.predict(x_test_scaled)
```

```python
y_test
```

```python
plt.scatter(x_test[:,0],x_test[:,1],c = y_pred)
```

### Other 2 ways to decide the number of clusters

```python
from kneed import KneeLocator
kl = KneeLocator(range(1,15),wcss,direction="decreasing",curve="convex")
kl.elbow
```

```python
silhoutee_coefficients = []
```

```python
from sklearn.metrics import silhouette_score
for k in range(2,15):
    kmeans = KMeans(n_clusters=k,init='k-means++',)
    kmeans.fit(x_train_scaled)
    score = silhouette_score(x_train_scaled,kmeans.labels_)
    silhoutee_coefficients.append(score)

```

```python
silhoutee_coefficients
```

```python
plt.plot(range(2,15), silhoutee_coefficients)
plt.xticks(range(2,15))
plt.xlabel("Number of clusters")
plt.ylabel("Silhoutee score")
```
